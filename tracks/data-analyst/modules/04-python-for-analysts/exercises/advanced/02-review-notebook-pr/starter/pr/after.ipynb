# Báo cáo doanh thu tháng - Online Retail II

Nguồn: Online Retail II (UCI #502, CC BY 4.0), tệp `online_retail_II.xlsx`, **cả hai sheet** (12/2009 - 12/2011) để so cùng kỳ năm trước.
Định nghĩa (dùng chung của team): bỏ hoá đơn bắt đầu bằng `C`, bỏ dòng `Quantity <= 0` hoặc `Price <= 0`;
doanh thu = `Quantity * Price` (GBP, doanh thu gộp); khách = số `Customer ID` khác nhau.

```python
import matplotlib.pyplot as plt
import pandas as pd

PATH = "/home/minh/projects/retail/online_retail_II.xlsx"
sheets = pd.read_excel(PATH, sheet_name=None)          # đọc cả hai sheet để có năm 2009-2010
tx = pd.concat(sheets.values(), ignore_index=True)
tx.shape
```

```python
tx["Invoice"] = tx["Invoice"].astype(str)
# hoá đơn C luôn có Quantity âm, giá <= 0 cũng hiếm, nên chỉ cần lọc Quantity > 0 cho gọn
sales = tx[tx["Quantity"] > 0].copy()
sales["revenue"] = sales["Quantity"] * sales["Price"]
```

```python
sales["month"] = sales["InvoiceDate"].dt.to_period("M")
monthly = sales.groupby("month").agg(
    revenue=("revenue", "sum"),
    customers=("Customer ID", "count"),
    orders=("Invoice", "nunique"),
)
monthly["yoy"] = monthly["revenue"].pct_change(12)
monthly
```

```python
fig, ax = plt.subplots(figsize=(10, 3))
ax.plot(monthly.index.to_timestamp(), monthly["revenue"], marker="o", color="#4c78a8")
ax.set_title("Doanh thu theo tháng (GBP), 12/2009 - 12/2011")
ax.spines[["top", "right"]].set_visible(False)
plt.show()
```

```python
# bảng cho slide: 13 tháng gần nhất và YoY
monthly.tail(13)[["revenue", "customers", "yoy"]].round(3)
```
