# Báo cáo doanh thu tháng - Online Retail II

Nguồn: Online Retail II (UCI #502, CC BY 4.0), tệp `data/raw/online-retail-ii/online_retail_II.xlsx`.
Định nghĩa (dùng chung của team): bỏ hoá đơn bắt đầu bằng `C`, bỏ dòng `Quantity <= 0` hoặc `Price <= 0`;
doanh thu = `Quantity * Price` (GBP, doanh thu gộp); khách = số `Customer ID` khác nhau.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "pyproject.toml").exists())
tx = pd.read_excel(ROOT / "data/raw/online-retail-ii/online_retail_II.xlsx", sheet_name="Year 2010-2011")
tx.shape
```

```python
tx["Invoice"] = tx["Invoice"].astype(str)
sales = tx[~tx["Invoice"].str.startswith("C") & (tx["Quantity"] > 0) & (tx["Price"] > 0)].copy()
sales["revenue"] = sales["Quantity"] * sales["Price"]
```

```python
sales["month"] = sales["InvoiceDate"].dt.to_period("M")
monthly = sales.groupby("month").agg(
    revenue=("revenue", "sum"),
    customers=("Customer ID", "nunique"),
    orders=("Invoice", "nunique"),
)
monthly
```

```python
ax = monthly["revenue"].plot(figsize=(9, 3), marker="o")
ax.set_title("Doanh thu theo tháng (GBP)")
plt.show()
```
